```python
import numpy as np

# Create an array of ones with 2 rows and 3 columns
arr1 = np.ones((2, 3))
print("Array of ones:")
print(arr1)

# Create an array of 7s with 2 rows and 2 columns
arr2 = np.full((2, 2), 7)
print("\nArray of 7s:")
print(arr2)

# Create even numbers from 2 to 10
arr3 = np.arange(2, 11, 2)
print("\nEven numbers:")
print(arr3)
```
