```python
import time, timeit, tracemalloc
N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)
def loop_double(values):
 result = []
 for x in values:
   result.append(x * 2)
 return result
def numpy_double(values):
 return values * 2
# Tool 1: perf_counter (one reading)
for label, func, data in (('loop', loop_double, py_list), ('numpy', numpy_double,
np_arr)):
 start = time.perf_counter()
 func(data)
 print(label, round(time.perf_counter() - start, 5), 'seconds')
# Tool 2: timeit (many runs, average)
n = 20
t_loop = timeit.timeit('loop_double(py_list)', globals=globals(), number=n) / n
t_np = timeit.timeit('numpy_double(np_arr)', globals=globals(), number=n) / n
print('loop', round(t_loop, 5), 'numpy', round(t_np, 5), 'ratio', round(t_loop /
t_np))
# Tool 3: tracemalloc (traced Python memory)
for label, make in (('list', lambda: [x * 2 for x in range(N)]),
 ('array', lambda: np.arange(N) * 2)):
 tracemalloc.start()
 obj = make()
 current, peak = tracemalloc.get_traced_memory()
 tracemalloc.stop()
 print(label, 'current MB', round(current / 1e6, 1), 'peak MB', round(peak / 1e6,
1)) 
```

```python
import numpy as np
import timeit
import tracemalloc
import sys
import os
import csv

def loop_version(data):
    result = []
    for x in data:
        result.append(x + 1)
    return result

def numpy_version(data):
    return data + 1

rows = []

for N in [1_000, 100_000, 1_000_000]:
    data = np.arange(N)
    number = 100 if N == 1_000 else 1

    t_loop = timeit.timeit(
        lambda: loop_version(data),
        number=number
    ) / number

    t_np = timeit.timeit(
        lambda: numpy_version(data),
        number=number
    ) / number

    ratio = t_loop / t_np
    rows.append((N, t_loop, t_np, ratio))

print(f"{'N':>10} {'Loop(s)':>12} {'NumPy(s)':>12} {'Ratio':>10}")

for N, t_loop, t_np, ratio in rows:
    print(f"{N:>10} {t_loop:>12.6f} {t_np:>12.6f} {ratio:>10.2f}")
```

```python
N = 1_000_000
data = np.arange(N)
n = 1

loop_times = timeit.repeat(
    lambda: loop_version(data),
    number=n,
    repeat=5
)

np_times = timeit.repeat(
    lambda: numpy_version(data),
    number=n,
    repeat=5
)

t_loop = min(loop_times) / n
t_np = min(np_times) / n

print("Best loop time:", t_loop)
print("Best NumPy time:", t_np)
print("Ratio:", t_loop / t_np)
```
