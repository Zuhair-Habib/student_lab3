```python

import numpy as np
import timeit
import tracemalloc
import os
import csv

# ==========================================
# PART 1: Compare Python Loop and NumPy Speed
# ==========================================

def loop_version(data):
    result = []
    for x in data:
        result.append(x + 1)
    return result


def numpy_version(data):
    return data + 1


rows = []

print("PART 1: Loop vs NumPy Timing")
print("-" * 65)
print(f"{'N':>12} {'Loop Time (s)':>15} {'NumPy Time (s)':>15} {'Ratio':>10}")

for N in [1_000, 100_000, 1_000_000]:
    data = np.arange(N)

    # Use 100 repetitions for small data; 1 for larger data
    number = 100 if N == 1_000 else 1

    t_loop = timeit.timeit(
        lambda: loop_version(data), number=number
    ) / number

    t_np = timeit.timeit(
        lambda: numpy_version(data), number=number
    ) / number

    ratio = t_loop / t_np if t_np != 0 else float("inf")

    rows.append([N, t_loop, t_np, ratio])

    print(f"{N:>12} {t_loop:>15.6f} {t_np:>15.6f} {ratio:>10.2f}x")


# ==========================================
# PART 2: Repeat Timing 5 Times for N=1 Million
# ==========================================

print("\nPART 2: Minimum Time from 5 Repetitions")
print("-" * 65)

N = 1_000_000
data = np.arange(N)
n = 1

loop_times = timeit.repeat(
    lambda: loop_version(data), number=n, repeat=5
)

numpy_times = timeit.repeat(
    lambda: numpy_version(data), number=n, repeat=5
)

t_loop_min = min(loop_times) / n
t_np_min = min(numpy_times) / n

print("Loop times:", loop_times)
print("NumPy times:", numpy_times)
print(f"Minimum loop time: {t_loop_min:.6f} seconds")
print(f"Minimum NumPy time: {t_np_min:.6f} seconds")
print(f"Speed ratio: {t_loop_min / t_np_min:.2f}x")

# Explanation:
# The minimum time helps reduce the effect of temporary
# background tasks and other system interruptions.


# ==========================================
# PART 3: Standardization Functions
# ==========================================

def standardise_loop(data):
    data = np.asarray(data, dtype=float)
    row_count, col_count = data.shape

    means = []
    stds = []

    # Calculate mean and standard deviation for each column
    for j in range(col_count):
        total = 0.0

        for i in range(row_count):
            total += data[i, j]

        mean = total / row_count
        means.append(mean)

        squared_total = 0.0

        for i in range(row_count):
            squared_total += (data[i, j] - mean) ** 2

        std = (squared_total / row_count) ** 0.5
        stds.append(std)

    result = np.zeros((row_count, col_count), dtype=float)

    # Standardize every value
    for i in range(row_count):
        for j in range(col_count):
            if stds[j] == 0:
                result[i, j] = 0.0
            else:
                result[i, j] = (
                    data[i, j] - means[j]
                ) / stds[j]

    return result


def standardise_np(data):
    data = np.asarray(data, dtype=float)

    # Calculate each column's mean and standard deviation
    means = data.mean(axis=0, keepdims=True)
    stds = data.std(axis=0, keepdims=True)

    # Avoid division by zero for constant columns
    safe_stds = np.where(stds == 0, 1.0, stds)

    # Broadcasting standardization
    return (data - means) / safe_stds


# ==========================================
# PART 4: Test Standardization Results
# ==========================================

print("\nPART 3: Standardization Test")
print("-" * 65)

rng = np.random.default_rng(0)
data = rng.random((20_000, 10))

loop_result = standardise_loop(data)
numpy_result = standardise_np(data)

print("Original data shape:", data.shape)
print("Loop result shape:", loop_result.shape)
print("NumPy result shape:", numpy_result.shape)
print("Both results are approximately equal:",
      np.allclose(loop_result, numpy_result))


# ==========================================
# PART 5: Compare Standardization Speed
# ==========================================

print("\nPART 4: Standardization Speed")
print("-" * 65)

t_loop = min(
    timeit.repeat(
        lambda: standardise_loop(data), number=1, repeat=5
    )
)

t_np = min(
    timeit.repeat(
        lambda: standardise_np(data), number=1, repeat=5
    )
)

standardise_ratio = t_loop / t_np if t_np != 0 else float("inf")

print(f"Loop standardization: {t_loop:.6f} seconds")
print(f"NumPy standardization: {t_np:.6f} seconds")
print(f"NumPy speed comparison: {standardise_ratio:.2f}x")


# ==========================================
# PART 6: Compare Peak Memory Usage
# ==========================================

print("\nPART 5: Peak Memory Usage")
print("-" * 65)

def measure_memory(func, data):
    tracemalloc.start()

    try:
        result = func(data)
        current, peak = tracemalloc.get_traced_memory()
    finally:
        tracemalloc.stop()

    del result
    return peak / (1024 ** 2)


loop_memory = measure_memory(standardise_loop, data)
numpy_memory = measure_memory(standardise_np, data)

print(f"Loop peak memory: {loop_memory:.2f} MB")
print(f"NumPy peak memory: {numpy_memory:.2f} MB")
print("Note: tracemalloc may not capture all NumPy native memory.")


# ==========================================
# PART 7: Save Results to CSV and NPY Files
# ==========================================

print("\nPART 6: Saving Files")
print("-" * 65)

# Create the output folder in the current working directory
os.makedirs("data/processed", exist_ok=True)

# Save timing results to CSV
csv_path = "data/processed/timing_results.csv"

with open(csv_path, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "Task",
        "N",
        "Loop_Time_Seconds",
        "NumPy_Time_Seconds",
        "Speed_Ratio"
    ])

    for N, loop_time, numpy_time, ratio in rows:
        writer.writerow([
            "Add One",
            N,
            loop_time,
            numpy_time,
            ratio
        ])

    writer.writerow([
        "Standardization",
        data.shape[0],
        t_loop,
        t_np,
        standardise_ratio
    ])

# Save standardized NumPy array to NPY
npy_path = "data/processed/standardised.npy"
np.save(npy_path, numpy_result)

print("CSV saved to:", csv_path)
print("NPY saved to:", npy_path)
print("\nAll tasks completed successfully!")
```
